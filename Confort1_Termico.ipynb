## Teoría del Confort Térmico en Ganado Vacuno

El confort térmico en el ganado vacuno se refiere a un estado de bienestar fisiológico en el que el animal no experimenta estrés por frío o calor. Cuando los animales se encuentran fuera de su zona de termoneutralidad (rango de temperatura en el que pueden mantener su temperatura corporal sin un gasto excesivo de energía), se produce estrés térmico, lo que puede afectar negativamente su productividad, salud y bienestar.

### Impacto del Estrés Térmico

El estrés por calor es una preocupación significativa en muchas regiones, especialmente en zonas tropicales y subtropicales. Sus efectos incluyen:

*   **Reducción de la producción de leche:** Las vacas lecheras disminuyen su consumo de alimento y, consecuentemente, su producción de leche. La calidad de la leche también puede verse afectada.
*   **Menor ganancia de peso:** El ganado de engorde reduce su consumo de alimento y su eficiencia de conversión, resultando en menores tasas de crecimiento.
*   **Problemas reproductivos:** El estrés por calor puede afectar la fertilidad en toros y vacas, aumentando los días abiertos y reduciendo las tasas de concepción.
*   **Disminución de la función inmunológica:** Los animales estresados térmicamente son más susceptibles a enfermedades.
*   **Cambios en el comportamiento:** Los animales pueden buscar sombra, reducir su actividad, aumentar la frecuencia respiratoria (jadeo) y buscar fuentes de agua para refrescarse.

### Factores que Influyen en el Confort Térmico

El confort térmico no depende solo de la temperatura del aire, sino de una combinación de factores ambientales y fisiológicos:

*   **Temperatura del Aire:** Directamente relacionada con el intercambio de calor entre el animal y su entorno.
*   **Humedad Relativa:** Alta humedad reduce la capacidad del animal para disipar calor por evaporación (sudoración y jadeo).
*   **Radiación Solar:** La exposición directa al sol aumenta la carga de calor radiante sobre el animal.
*   **Velocidad del Viento:** Una brisa puede ayudar a disipar el calor del cuerpo, pero un viento excesivo en condiciones frías puede aumentar la pérdida de calor.
*   **Temperatura de la Superficie:** La temperatura del suelo o de las superficies circundantes puede influir en la carga de calor radiante.
*   **Características del Animal:** Raza (razas europeas son más susceptibles al calor que las cebuínas), edad, estado fisiológico (producción de leche, gestación), color del pelaje y condición corporal influyen en su tolerancia al estrés térmico.

### Índices de Confort Térmico

Para cuantificar el grado de estrés térmico, se utilizan índices que combinan varios parámetros meteorológicos. El Índice de Temperatura y Humedad (ITH) es uno de los más comunes y se utiliza para evaluar el estrés por calor en el ganado. Otros índices más complejos pueden incluir la velocidad del viento y la radiación solar. Estos índices permiten a los productores tomar decisiones informadas sobre estrategias de mitigación para proteger a sus animales del estrés térmico.

# *Métodos y cálculos*

- Índice de temperatura y humedad (producción intensiva)

$$ITH = 0.72 * (T_{bs} + T_{bh}) + 40.6$$

Donde:

1.  $T_{bs}$ Temperatura de Bulbo seco ($^oC$)

2. $T_{bh}$ Temperatura de bulbo húmedo ($^oC$)

Para la correlación de los datos de la especie bovina en la producción de leche se recomienda utilizar los límites propuestos por National Weather Service:

1. Sí $ITH <= 70$ ambiente no estresante.

2. Sí $ITH > 70 \space And \space ITH <= 78$ situación de alerta por estrés térmico.

3. Sí $ITH > 78 \space And \space ITH <= 83$ peligro por estrés térmico .

4. Sí $ITH > 83$ situación crítica.

````python
# ENTRADA: Pedir tbs (Temperatura de bulbo secoo) y tbh (Temperatura de bulbo húmedo)
tbs = float(input("Ingrese la temperatura de bulbo seco, Grados Centigrados: "))
tbh = float(input("Ingrese la temperatura de bulbo húmedo, Grados Centigrados: "))

# PROCESO: Calculo de ith (índice de Temperatura y Humedad)
ith = 0.72 * (tbs + tbh) + 40.6

# SALIDA: Imprimir resultado de la evaluación del ith y el confort térmico
if ith <= 70:
  print("El índice de temperatura y humedad es: ", ith)
  print("Ambiente no estresante")

elif ith > 70 and ith <= 78:
  print("El índice de temperatura y humedad es: ", ith)
  print("Situación de alerta por estrés térmico")

elif ith > 78 and ith <= 83:
  print("El índice de temperatura y humedad es: ", ith)
  print("peligro por estrés térmico")

elif ith > 83:
  print("El índice de temperatura y humedad es: ", ith)
  print("Situación crítica")
````
