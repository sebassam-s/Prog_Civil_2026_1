# Problema de Ingeniería (Nivel Básico)

Un ingeniero agrícola necesita calcular **el volumen de agua de un tanque cilíndrico** utilizado para riego.

La fórmula del volumen de un cilindro es:

[
V = \pi \cdot r^2 \cdot h
]

donde:

* ( r ) = radio del tanque (m)
* ( h ) = altura del tanque (m)
* ( V ) = volumen del tanque (m³)

---

# 1️⃣ Definir Entradas, Proceso y Salidas

## Entradas

Datos que el usuario debe ingresar:

* **radio** del tanque (m)
* **altura** del tanque (m)

## Proceso

1. Elevar el radio al cuadrado
2. Multiplicar por π
3. Multiplicar por la altura

Matemáticamente:

[
volumen = \pi \times radio^2 \times altura
]

## Salida

* **Volumen del tanque en metros cúbicos**

---

# 2️⃣ Algoritmo (explicado paso a paso)

1. Solicitar al usuario el **radio del tanque**
2. Solicitar al usuario la **altura del tanque**
3. Calcular el volumen usando la fórmula del cilindro
4. Mostrar el resultado en pantalla

---

# 3️⃣ Pseudocódigo

```text
Inicio

    Escribir "Ingrese el radio del tanque:"
    Leer radio

    Escribir "Ingrese la altura del tanque:"
    Leer altura

    volumen = π * radio^2 * altura

    Escribir "El volumen del tanque es:", volumen

Fin
```

---

# 4️⃣ Código en Python

```python
# Programa para calcular el volumen de un tanque cilíndrico
# Aplicación básica para ingeniería agrícola

import math  # Librería para usar el valor de pi

# 1. Entradas
radio = float(input("Ingrese el radio del tanque (m): "))
altura = float(input("Ingrese la altura del tanque (m): "))

# 2. Proceso
volumen = math.pi * radio**2 * altura

# 3. Salida
print("El volumen del tanque es:", volumen, "m³")
```

### Un ejemplo sencillo: El cobrador de entradas

Imagina que queremos hacer un programa que pregunte la edad a una persona y le diga si puede pasar a ver una película de terror.

```python
# Así se ve un código en Python

nombre = input("¿Cómo te llamas? ")
edad = int(input("¿Cuántos años tienes? "))

if edad >= 18:
    print(f"¡Bienvenido, {nombre}! Puedes pasar a la sala.")
else:
    años_faltantes = 18 - edad
    print(f"Lo siento, {nombre}. Vuelve en {años_faltantes} años.")

```

**¿Qué está pasando aquí?**

1. `input`: Le pide información al usuario.
2. `int`: Convierte el texto que escribes en un número para poder hacer cálculos.
3. `if / else`: Es la lógica de decisión ("Si pasa esto, haz aquello; si no, haz esto otro").
4. `f"..."`: Es una forma elegante de mezclar texto con variables.
